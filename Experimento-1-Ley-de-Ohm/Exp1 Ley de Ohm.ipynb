# Experimento 1: Ley de Ohm

Este notebook presenta el análisis de datos experimentales obtenidos en laboratorio
para el estudio de la relación entre voltaje y corriente en resistores eléctricos.

El objetivo es verificar experimentalmente la Ley de Ohm mediante ajuste lineal,
utilizando herramientas de análisis de datos y programación científica en Python.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

```

### 1. Análisis de los datos para el tungsteno

```python
data1 = pd.read_excel("tungsteno.xlsx")

V1 = data1["V (Voltios)"]   # Voltaje (V)
I1 = data1["A (Amperios)"]   # Corriente (A)
```

```python
data1
```

```python
# Realizamos el ajuste lineal
slope1, b1, r_value1, p_value1, std_err1 = stats.linregress(V1, I1)

m1 = slope1 # pendiente
m1_error = std_err1 # error de la pendiente
```

```python
# Cálculo de la resistencia (R)
R1 = 1 / m1
R1_error = m1_error / (m1**2)
```

```python
# Gráfica de voltaje vs corriente

V1_fit = np.linspace(min(V1), max(V1), 100)
I1_fit = m1 * V1_fit + b1

plt.figure()
plt.scatter(V1, I1, label="Datos experimentales")
plt.plot(V1_fit, I1_fit, label="Ajuste lineal", color="orange")

plt.xlabel("Voltaje (V)")
plt.ylabel("Corriente (A)")
plt.title("Curva Voltaje–Corriente")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Resultados del ajuste lineal:")
print(f"Pendiente:  m = {m1:.6f} ± {m1_error:.6f} A/V")
print(f"Intercepto: b = {b1:.4e} A")
print(f"Coeficiente de determinación R²: {r_value1**2:.5f}")
print(f"Resistencia: R = ({R1:.4f} ± {R1_error:.4f}) Ω")
```

```python
# Gráfica con los errores

# Error instrumental de la corriente

error_instrumental = 0.01
I1_error = np.full_like(I1, error_instrumental)


plt.figure()
plt.errorbar(
    V1,
    I1,
    yerr=I1_error,
    fmt='o',
    capsize=5,
    label="Datos experimentales"
)

plt.plot(
    V1_fit,
    I1_fit,
    label=f"Ajuste: I = ({m1:.4f} ± {m1_error:.4f}) V + {b1:.2e}"
)

plt.xlabel("Voltaje (V)")
plt.ylabel("Corriente (A)")
plt.title("Corriente vs Voltaje – Carbón")
plt.legend()
plt.grid(True)
plt.show()
```

### 3. Análisis de los datos para el diodo (polaridad directa)

```python
data3 = pd.read_excel("diodo_directa.xlsx")

V3 = data3["V (Voltios)"]   # Voltaje (V)
I3 = data3["A (Amperios)"]   # Corriente (A)
```

```python
data3
```

```python
# Filtrar valores positivos (log no acepta cero o negativos)
mask = I3 > 0
V3 = V3[mask]
I3 = I3[mask]

# Linealización
lnI = np.log(I3)

# Ajuste lineal
slope3, b3, r_value3, p_value3, std_err3 = stats.linregress(V3, lnI)

a = slope3
I0 = np.exp(b3)

print("========== AJUSTE EXPONENCIAL ==========")
print(f"a = {a:.6e}")
print(f"I0 = {I0:.6e}")
print(f"R² = {r_value3**2:.5f}")

# Curva ajustada
V3_fit = np.linspace(min(V3), max(V3), 100)
I3_fit = I0 * np.exp(a * V3_fit)

plt.figure()
plt.scatter(V3, I3, label="Datos experimentales")
plt.plot(V3_fit, I3_fit, label="Ajuste exponencial", color="purple")

plt.xlabel("Voltaje (V)")
plt.ylabel("Corriente (A)")
plt.title("Diodo - Ajuste Exponencial")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("\n========== FUNCIÓN AJUSTADA ==========")
print(f"I(V) = {I0:.3e} * exp({a:.3f} * V)")
```

Notemos que los datos tienden a formar una función exponencial, por lo que no es conveniente aplicar una regresión lineal.
